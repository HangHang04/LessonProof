```python
import csv
import json

with open('loans.csv', encoding='utf-8', newline='') as source:
    rows = list(csv.DictReader(source))
```

```python
result = {
    'book_count': len(rows),
    'total_borrows': sum(int(row['borrow_count']) for row in rows),
}
print(json.dumps(result, sort_keys=True))
```
