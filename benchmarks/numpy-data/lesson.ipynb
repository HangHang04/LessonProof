```python
import json
import numpy

data = numpy.loadtxt(fname='data/inflammation-01.csv', delimiter=',')
```

```python
print(json.dumps({'shape': list(data.shape), 'first_value': int(data[0, 0]), 'middle_value': int(data[29, 19])}, sort_keys=True))
```

```python
print(json.dumps({'minimum': int(numpy.min(data)), 'maximum': int(numpy.max(data)), 'mean_5dp': format(float(numpy.mean(data)), '.5f')}, sort_keys=True))
```
